```python
#1 minimum convey capacity
# parameters :- N - how many , D = how many runs , and the array
index = 0
def mcc(n,d,arr,num):
    def convey(c):
        print("--------------------------- \n current c: ",c)
        val = 0
        runs = 1
        print("d:",d)
        index = 0
        for i in range(n):
            
            if (val + arr[i] ) <= c :
                print(val," = ",val ,"+", arr[i] ," = ",val + arr[i])
                val = val + arr[i]
                index = i
            #if (val + arr[i] ) >= c :
                #index = i
            if i < n-1 and val + arr[i+1]  > c and c>arr[i]:
                runs = runs + 1
                print("Runs:",runs)
                val = 0           
        return runs,index

    for j in range(1,num):
        res,index = convey(j)
        print("Runs: ",res)
        if res==d and index == n-1:
            print("Result ------>",j,"index:",index)
            return
            
            

arr = [int(x) for x in input("Enter the packages:").split(" ")]
d = int(input("Enter max runs in a day:"))
print(mcc(len(arr),d,arr,100000))
```

```python
#3 attendence  -  attendance is stored in binary ,1 present, 0 absent , now ther i k whihc is the forgive, 
#110011001110 , k=2 , then mbest window is 1100111 - the two absents are forgiven.

def attendance(n,k,arr):
    li = []
    binary = ""
    count = 0
    for i in range(len(arr)):
        for j in range(i,len(arr)):
            if count<k:
                if arr[j]!= 0:
                    binary += str(arr[j])
                else:
                    binary += str(arr[j])
                    count = count + 1
            if count==k:
                break
        li.append(int(binary))
    return li
            
            

arr = [int(x) for x in input("Enter attendence:").split(" ")]
k = int(input("enter the k:"))
print(attendance(len(arr),k,arr)) 
```
